```python
import pandas as pd
import numpy as np
import sqlite3

ocorrencia = pd.read_csv('analise-acidentes-aereos-cenipa/data/ocorrencia.csv', encoding='latin-1', sep=';')
aeronave = pd.read_csv('analise-acidentes-aereos-cenipa/data/aeronave.csv', encoding='latin-1', sep=';')
ocorrencia_tipo = pd.read_csv('analise-acidentes-aereos-cenipa/data/ocorrencia_tipo.csv', encoding='latin-1', sep=';') # carregando as 3 tabelas

con = sqlite3.connect('db') # criação de banco de dados local usando sqlite3

ocorrencia.to_sql(name = "ocorrencia",con = con, if_exists =  'replace')   # colocando os 3 dataframes no banco de dados SQL
aeronave.to_sql(name = "aeronave",con = con, if_exists = 'replace')
ocorrencia_tipo.to_sql(name = "ocorrencia_tipo",con = con, if_exists ='replace')
```
