# Investigando uma diferença de linhas após JOIN

Quando juntamos as tabelas *ocorrencia* e *aeronave*, o resultado tem **mais** linhas do que ocorrencia **sozinha**, o que nos leva a pensar, "nossa, será que tem alguma ocorrência sendo duplicada?"

```python
query_sem_join = """SELECT COUNT(*) FROM ocorrencia"""
pd.read_sql_query(query_sem_join, con)
```

```python
query_com_join = """SELECT COUNT(*) FROM ocorrencia JOIN aeronave ON aeronave.codigo_ocorrencia2 = ocorrencia.codigo_ocorrencia2"""
pd.read_sql_query(query_com_join, con)
```

Porém, ao analisar a coluna "total_aeronaves_envolvidas" da tabela ocorrencia, podemos perceber que existem casos em que mais de uma aeronave está associada a um único acontecimento.

```python
query_aeronave_env = """ SELECT total_aeronaves_envolvidas FROM ocorrencia WHERE total_aeronaves_envolvidas > 1"""
pd.read_sql_query(query_aeronave_env, con)
```

Os 100 casos com múltiplas aeronaves parecem explicar a diferença de linhas, mas será que explicam exatamente? Se a soma de total_aeronaves_envolvidas em todas as ocorrências bater com o total de linhas do JOIN (9926), confirma que não sobra nada sem explicação.

```python
aeronaves_envolvidas = """ SELECT SUM(total_aeronaves_envolvidas) FROM ocorrencia"""
pd.read_sql_query(aeronaves_envolvidas, con)
```

A soma bateu exatamente: 9926. Isso confirma que a diferença de linhas após o JOIN se explica inteiramente pelas ocorrências com múltiplas aeronaves, não há duplicação nem perda de dados no processo.

Mesmo sabendo que as linhas estão corretas, a coluna aeronave_fatalidades_total é per-aeronave ou total repetido?

```python
per_aeronave = """ SELECT codigo_ocorrencia2,aeronave_fatalidades_total, aeronave_matricula FROM aeronave WHERE codigo_ocorrencia2 = 46255 """
pd.read_sql_query(per_aeronave, con)
```

Cada linha da coluna aeronave_fatalidades_total registra as fatalidades daquela aeronave específica, não o total da ocorrência repetido. Na ocorrência 46255, a PTNIR teve 2 fatalidades e a PTNKA teve 0. Por isso, somar essa coluna após o JOIN não conta a mesma morte duas vezes, e o total por ano é confiável

##Quais tipos de ocorrência são mais letais?

Antes de partir para a checagem em si, vamos analisar se a ocorrencia_tipo tem uma linha por ocorrência: Escolhi utilizar elas porque elas guardam, respectivamente, as colunas ocorrencia_tipo e aeronave_fatalidades_total.

```python
ocorrencia_count = """SELECT COUNT(*), COUNT(DISTINCT codigo_ocorrencia1) FROM ocorrencia_tipo"""
pd.read_sql_query(ocorrencia_count, con)
```

```python
ocorrencia_rep = """ SELECT COUNT(codigo_ocorrencia1), codigo_ocorrencia1
FROM ocorrencia_tipo
GROUP BY codigo_ocorrencia1
HAVING COUNT(*) > 1 """

pd.read_sql_query(ocorrencia_rep, con)
```

```python
ocorr_31255 = """ SELECT codigo_ocorrencia1, ocorrencia_tipo
FROM ocorrencia_tipo
WHERE codigo_ocorrencia1 = 31255"""

pd.read_sql_query(ocorr_31255, con)
```

Com essas buscas, podemos notar que o número de linhas da tabela ocorrencia_tipo é diferente do número de ocorrências distintas pois existem tipos de ocorrência diferentes que estão atribuidos a um mesmo código de ocorrência, mesmo evento. Na busca pela relação de fatalidade <-> tipo de ocorrência, poderia ocorrer erro com fatalidades sendo contadas mais de uma vez

**O que fazer com ocorrências de vários tipos?** Nessa análise, eu escolhi englobar ambos os tipos envolvidos em uma ocorrência, ou seja, ela não vai ser categorizada apenas como "Perda de controle no solo" ou "Pouso longo", mas sim os dois, porque são detalhes que não se separam.

**O que significa mais "letal" nesse contexto?** Aqui, escolhi utilizar a proporção de acidentes fatais por tipo de ocorrência, pois num acidente com um grande número de fatalidades, não teríamos uma inflação numérica na contagem de mortes (que pode acabar aumentando a média do tipo)

```python
fatais_tipo = """WITH mortes_por_ocorrencia AS ( SELECT codigo_ocorrencia2, sum(aeronave_fatalidades_total) AS total_mortes
FROM aeronave
GROUP BY codigo_ocorrencia2 )

SELECT ocorrencia_tipo, COUNT(*) AS total_ocorrencias, SUM(CASE WHEN total_mortes >= 1 THEN 1 ELSE 0 END) AS ocorrencias_fatais
FROM ocorrencia_tipo
JOIN mortes_por_ocorrencia ON ocorrencia_tipo.codigo_ocorrencia1 = mortes_por_ocorrencia.codigo_ocorrencia2
GROUP BY ocorrencia_tipo"""
pd.read_sql_query(fatais_tipo, con)

```
