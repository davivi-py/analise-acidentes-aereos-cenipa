# Investigando uma diferença de linhas após JOIN

Quando juntamos as tabelas *ocorrencia* e *aeronave*, o resultado tem **mais** linhas do que ocorrencia **sozinha**, o que nos leva a pensar, "nossa, será que tem alguma ocorrência sendo duplicada?"

```python
query_sem_join = """SELECT COUNT(*) FROM ocorrencia"""
pd.read_sql_query(query_sem_join, con)
```

```python
query_com_join = """SELECT COUNT(*) FROM ocorrencia JOIN aeronave ON aeronave.codigo_ocorrencia2 = ocorrencia.codigo_ocorrencia2"""
pd.read_sql_query(query_com_join, con)
```

Porém, ao analisar a coluna "total_aeronaves_envolvidas" da tabela ocorrencia, podemos perceber que existem casos em que mais de uma aeronave está associada a um único acontecimento.

```python
query_aeronave_env = """ SELECT total_aeronaves_envolvidas FROM ocorrencia WHERE total_aeronaves_envolvidas > 1"""
pd.read_sql_query(query_aeronave_env, con)
```

Os 100 casos com múltiplas aeronaves parecem explicar a diferença de linhas, mas será que explicam exatamente? Se a soma de total_aeronaves_envolvidas em todas as ocorrências bater com o total de linhas do JOIN (9926), confirma que não sobra nada sem explicação.

```python
aeronaves_envolvidas = """ SELECT SUM(total_aeronaves_envolvidas) FROM ocorrencia"""
pd.read_sql_query(aeronaves_envolvidas, con)
```

A soma bateu exatamente: 9926. Isso confirma que a diferença de linhas após o JOIN se explica inteiramente pelas ocorrências com múltiplas aeronaves, não há duplicação nem perda de dados no processo.
